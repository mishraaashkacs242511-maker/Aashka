```python
import pandas as pd
import tensorflow as tf
import matplotlib.pyplot as plt

df = pd.read_csv("store_sales_data.csv")

X = df[["Quantity", "Discount"]].values
y = df["Sales"].values

model = tf.keras.Sequential([
    tf.keras.layers.Dense(10, activation="relu", input_shape=(2,)),
    tf.keras.layers.Dense(10, activation="relu"),
    tf.keras.layers.Dense(1)
])

model.compile(optimizer="adam", loss="mse")

history = model.fit(X, y, epochs=30, verbose=0)

print("Model trained successfully!")

plt.figure(figsize=(8,5))
plt.plot(history.history["loss"], marker="o")
plt.title("TensorFlow Training Loss")
plt.xlabel("Epoch")
plt.ylabel("Loss")
plt.grid(True)
plt.show()

region_sales = df.groupby("Region")["Sales"].sum()

plt.figure(figsize=(8,5))
region_sales.plot(kind="bar")
plt.title("Total Sales by Region")
plt.xlabel("Region")
plt.ylabel("Total Sales")
plt.xticks(rotation=0)
plt.grid(axis="y")
plt.show()

category_profit = df.groupby("Category of Goods")["Profit"].sum()

plt.figure(figsize=(10,5))
category_profit.plot(kind="bar")
plt.title("Total Profit by Category")
plt.xlabel("Category of Goods")
plt.ylabel("Total Profit")
plt.xticks(rotation=45, ha="right")
plt.grid(axis="y")
plt.show()
```
